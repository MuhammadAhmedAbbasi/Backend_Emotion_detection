# Start Service
Run the cells in order. The last cell starts Uvicorn using the notebook's existing asyncio event loop.

```python
from pathlib import Path
NB_DIR = Path.cwd() if Path.cwd().name == "Service_jupyter" else Path.cwd() / "Service_jupyter"
get_ipython().run_line_magic("run", str(NB_DIR / "app.ipynb"))

HOST = "127.0.0.1"
PORT = 5001

print("Application loaded.")
print(f"Dashboard: http://{HOST}:{PORT}")
print(f"API docs:  http://{HOST}:{PORT}/docs")
```

```python
import uvicorn

# Jupyter already has an asyncio event loop running.
# Therefore we create a Uvicorn Server and await it instead of using uvicorn.run().
config = uvicorn.Config(app, host=HOST, port=PORT, reload=False, log_level="info")
server = uvicorn.Server(config)

print(f"Starting service at http://{HOST}:{PORT}")
print("Interrupt this cell to stop the service.")

await server.serve()
```
